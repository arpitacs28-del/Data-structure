```python
#write mode
file = open("data.txt","w")#open file in write mode
file.write("this is a new line of text.\n")
file.close()
```

```python
#reading a file
file = open("data.txt","r")#open file in read mode
content = file.read()
print(content)
file.close()
```

```python
#write mode
file = open("data.txt","w")#open file in write mode
file.write("hello world.\n")
file.close()
```

```python
#create mode
file = open("data.txt","x")# open file in create mode
file.close()
```

```python
#append mode
file = open("data.txt","a") #open file in append mode
file.write("this is an append line of text\n")
file.close()
```

```python
#file automatically closes after the block of code
with open("data.txt","r") as file:
    content = file.read()
    print(content)
```

```python
#read specific number of character
file = open("data.txt","r")
print(file.read(10))#read firat 10 characters
file.close()
```

```python
#read one line at a time
file = open("data.txt","r")
print(file.readline())#read first line
file.close()
```

```python
file = open("data.txt","r")
print(file.name)#name of the file
print(file.mode)#mode in which the is opened
print(file.closed)#check if the file is closed or not
print(file.read(5))
file.read(5)
print(file.tell())#current position of the file cursor
file.seek(0)#move the file cursor to the begining of the file
```

```python
try:
    file = open('newfile.txt','x')
    print("file created successfully")
except FileExistsError:
    print("File already exists")
finally:
    file.close()    
```
